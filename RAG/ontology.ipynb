```python
from owlready2 import *
```

```python
# Create new ontology
onto = get_ontology("http://test.org/onto.owl")
with onto:
    # Define Classes
    class Patient(Thing):
        pass

    class Disease(Thing):
        pass

    class Symptom(Thing):
        pass

    class Treatment(Thing):
        pass

    class Medication(Thing):
        pass

    class Doctor(Thing):
        pass

    # Define Data Properties
    class hasName(DataProperty):
        domain = [Patient, Disease, Treatment, Medication, Doctor]
        range = [str]

    class hasAge(DataProperty):
        domain = [Patient]
        range = [int]

    class hasGender(DataProperty):
        domain = [Patient]
        range = [str]

    class hasDescription(DataProperty):
        domain = [Symptom]
        range = [str]

    class hasDosage(DataProperty):
        domain = [Medication]
        range = [str]

    class hasSpecialization(DataProperty):
        domain = [Doctor]
        range = [str]

    # Define Object Properties
    class hasDisease(ObjectProperty):
        domain = [Patient]
        range = [Disease]

    class hasSymptom(ObjectProperty):
        domain = [Disease]
        range = [Symptom]

    class treatedBy(ObjectProperty):
        domain = [Disease]
        range = [Treatment]

    class includesMedication(ObjectProperty):
        domain = [Treatment]
        range = [Medication]

    class prescribedFor(ObjectProperty):
        domain = [Medication]
        range = [Disease]

    class treats(ObjectProperty):
        domain = [Doctor]
        range = [Patient]

# Save the ontology to a file
onto.save(file = "med_ontology.owl", format = "rdfxml")

# You can also view the ontology in the console
for cls in onto.classes():
    print(cls)

for prop in onto.properties():
    print(prop)
```

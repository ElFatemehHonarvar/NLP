# Customize NER

```python
# default

import spacy

nlp = spacy.load("en_core_web_sm")

doc = nlp("Mike is going to acquire Seven for $45 billion and sale Mike brand")

for ent in doc.ents:
  print(ent.text, "|", ent.label_)
```

```python
print(doc[1:3], ", type : ", type(doc[1:3]))
```

```python
from spacy.tokens import Span


s1 = Span(doc, 5, 6, label="ORG")
s2 = Span(doc, 12, 13, label= "ORG")

doc.set_ents([s1, s2], default="unmodified")

for ent in doc.ents:
  print(ent.text, "|", ent.label_)
```

## Word Vectors and spaCy
***Note: The English small model does not have word vectors or word embeddings saved however the medium  model does.***

Initial methods for creating word vectors in a pipeline take all words in a corpus and convert them into a single, unique number. These words are then stored in a dictionary that would look like this: {“the”: 1, “a”, 2} etc. This is known as a *bag of words*. This approach to representing words numerically, however, only allow a computer to understand words numerically to identify unique words. It does not, however, allow a computer to understand meaning. Hence, for text understanding or trying to get a computer system to be able to understand how a word functions within the sentence and how relates to other words in general that does not really work for us.

***List of all supported Named Entities***
- PERSON:      People, including fictional.
- NORP:        Nationalities or religious or political groups.
- FAC:         Buildings, airports, highways, bridges, etc.
- ORG:         Companies, agencies, institutions, etc.
- GPE:         Countries, cities, states.
- LOC:         Non-GPE locations, mountain ranges, bodies of water.
- PRODUCT:     Objects, vehicles, foods, etc. (Not services.)
- EVENT:       Named hurricanes, battles, wars, sports events, etc.
- WORK_OF_ART: Titles of books, songs, etc.
- LAW:         Named documents made into laws.
- LANGUAGE:    Any named language.
- DATE:        Absolute or relative dates or periods.
- TIME:        Times smaller than a day.
- PERCENT:     Percentage, including ”%“.
- MONEY:       Monetary values, including unit.
- QUANTITY:    Measurements, as of weight or distance.
- ORDINAL:     “first”, “second”, etc.
- CARDINAL:    Numerals that do not fall under another type.

## Matcher
https://spacy.io/api/matcher

How the data is extracted is the key difference between Matcher and EntityRuler. Matcher stores information within the vocab of the NLP model. it stores as a unique identifier or lexeme and is not going to store in the doc.ents.
- *Entity ruler*: the thing you are trying to extract is sth that is important to have a label that corresponds to it within the entities that are coming out. ex: if you are working with finances, or the names of streets.
- *Matcher*: you are looking for sth that is not necessarily an entity type, but sth that is structured within the text that will help you extract information. Matcher is not matching a sequence of characters but matching a sequence of linguistic features. Ex: find all instances of Proper noun + Verb.
all patterns in Matcher can also implemented in EntityRuler

- *Regex*: you have a complicated pattern you need to extract and that pattern is not dependent upon a specific part of speech.
In fact, Regex is just a string-matching framework and it is not a framework for identifying linguistic features using them, and extracting them

````python
import spacy
from spacy.matcher import Matcher
````

<font color="green" >***A pipeline component is a function that receives a Doc object, modifies it and returns***</font>
 

````python
#Now let's remove GPE lable
@Language.component("remove_gpe")

def remove_gpe(doc):

    original_ent = list(doc.ents)
    for ent in doc.ents:
        if ent.label_ == "GPE":
           original_ent.remove(ent)

    doc.ents = original_ent
    return doc
````

````python
import spacy
from spacy.language import Language
from spacy.tokens import Span

@Language.component("add_ner")

def add_ner(doc):
    original_ents = list(doc.ents)
    pattern = r"Sam [A-Z]\w+"
    matches = re.finditer(pattern, text)
    for match in matches:
        strat_chr, end_char = match.span()
        span = doc.char_span(strat_chr, end_char, "PERSON")
        original_ents.append(Span(doc, span.start, span.end, span.label_))
    doc.ents = original_ents    
    return doc

````

***filter_spans***:
Filter a sequence of Span objects and remove duplicates or overlaps. Useful for creating named entities (where one token can only be part of one entity) or when merging spans with Retokenizer.merge. ***When spans overlap, the (first) longest span is preferred over shorter spans.***
Link: https://spacy.io/api/top-level

````python
import spacy
import re
from spacy.language import Language
from spacy.tokens import Span
from spacy.util import filter_spans

@Language.component("add_ner")

def add_ner(doc):
    original_ents = list(doc.ents)
    pattern = r"Hollywood"
    matches = re.finditer(pattern, text)
    for match in matches:
        strat_chr, end_char = match.span()
        span = doc.char_span(strat_chr, end_char, "CINEMA")
        original_ents.append(Span(doc, span.start, span.end, span.label_))
    filtered = filter_spans(original_ents)
    doc.ents = filtered 
    return doc
````

````python
nlp = spacy.load("en_core_web_sm")
nlp.add_pipe("add_ner")
text = "Sam Kidman was a teacher, but Sam Hollywood is a TV Host. The name Sam is so common."

doc = nlp(text)

for ent in doc.ents:
    print(ent.text, ent.label_)

# Sam Hollywood is longer token than Hollywood So CINIMA does not come out.
````

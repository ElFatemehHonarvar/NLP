- > class1: 1200
- > class2: 2500

class2 => select 1200 sample randomly

class1 => copy 2500-1200 sample : ``class1.sample(class2, replace=True)``  so it duplicate the samples to reach class2.
Pandas ``sample()`` is a method used to randomly select rows or columns ``from a DataFrame or Series.

## Let's create a custom comonent

```python
import spacy
source_nlp = spacy.load("en_core_web_sm")
```

```python
nlp = spacy.blank("en")
print(nlp.pipe_names)

text = "Iranian do not have access to the free internet while people from all over the world use Twitter and Telegram."
doc = nlp(text)
for ent in doc.ents:
  print(f" entity : {ent.text}, label : {ent.label_}")
```

```python
nlp = spacy.blank("en")
nlp.add_pipe("ner", source=source_nlp)  # the source is from other trained pipeline
nlp.pipe_names
```

```python
text = "Iranian do not have access to the free internet while people from all over the world use Twitter and Telegram."
doc = nlp(text)
for ent in doc.ents:
  print(f" entity : {ent.text}, label : {ent.label_}")
```

## Let's dive into Stemming and Lemmatization

Note : Stemmer does not have the knowledge of the language and just using fix rules. However, it s faster than Lem.

> spaCy does not support stemming so we use nltk

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["running", "ate", "eat", "went", "capable", "useful", "going", "better"]
for word in words:
  print(f" word : {word} |  stemmed word : {stemmer.stem(word)}")
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")  # differnt language models have differnt rules for lemmatization

text = " ".join(words)
doc = nlp(text)
for token in doc:
  print(f"word : {token} |  lemma : {token.lemma_} | hash : {token.lemma}")
```

Attribute rulers -session 10

Modify a behavior(customize): Attribute rulers assigns attribute to a particular token : me : for example when we use slang language the model may fail to recognize lemma correctly

```python
text = "Bro, you wanna go? Brah, don't say no! I am exhausted! yeah I am "
doc = nlp(text)
for token in doc:
  print(f"word : {token} | lemma : {token.lemma_}")
```

```python
print(nlp.pipe_names)
```

```python
att_ruler = nlp.get_pipe("attribute_ruler")
att_ruler.add([[{"TEXT": "Bro"}], [{"TEXT": "Brah"}]], {"LEMMA":'brother'})
att_ruler.add([[{"TEXT": "yeah"}]], {"LEMMA": "yes"})

doc = nlp(text)
for token in doc:
  print(f"word : {token} | lemma : {token.lemma_}")
```

To create a NLP application, you may want to remove punctuation marks, space, and other

```python
text = """Microsoft Corp. today announced the following results for the quarter ended December 31, 2021, as compared to the corresponding period of last fiscal year:

·         Revenue was $51.7 billion and increased 20%
·         Operating income was $22.2 billion and increased 24%
·         Net income was $18.8 billion and increased 21%
·         Diluted earnings per share was $2.48 and increased 22%
“Digital technology etc. is the most malleable resource at the world’s disposal to overcome constraints and reimagine everyday work and life,” said Satya Nadella, chairman and chief executive officer of Microsoft. “As tech as a percentage of global GDP continues to increase, we are innovating and investing across diverse and growing markets, with a common underlying technology stack and an operating model that reinforces a common strategy, culture, and sense of purpose.”
“Solid commercial execution, represented by strong bookings growth driven by long-term Azure commitments, increased Microsoft Cloud revenue to $22.1 billion, up 32% year over year” said Amy Hood, executive vice president and chief financial officer of Microsoft."""
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)
for token in doc:
  print(f"token : {token},  pos: {token.pos_} | {spacy.explain(token.pos_)} | {token.tag_} | {spacy.explain(token.tag_)}")
```

```python
filtered_tokens = []
for token in doc:
  if token.pos_ in ["SPACE", "X", "PUNCT"]:
    filtered_tokens.append(token)
  else:
    print(f"token : {token},  pos: {token.pos_} | {spacy.explain(token.pos_)}")
```

Count the number of verbs, nouns, etc.

```python
count = 0
for token in doc:
  if token.pos_ in ["NOUN", "PROPN"]:
    count +=1
print(f" there are {count} noun and pronoun in this text")
```

Spacy makes this easier by doc.count_by

```python
dic_count = doc.count_by(spacy.attrs.POS)

for key, value in dic_count.items():
  print(f"{doc.vocab[key].text} :{value}")
```

## Let's create a custom comonent

```python
import spacy
source_nlp = spacy.load("en_core_web_sm")
```

```python
nlp = spacy.blank("en")
print(nlp.pipe_names)

text = "Iranian do not have access to the free internet while people from all over the world use Twitter and Telegram."
doc = nlp(text)
for ent in doc.ents:
  print(f" entity : {ent.text}, label : {ent.label_}")
```

```python
nlp = spacy.blank("en")
nlp.add_pipe("ner", source=source_nlp)  # the source is from other trained pipeline
nlp.pipe_names
```

```python
text = "Iranian do not have access to the free internet while people from all over the world use Twitter and Telegram."
doc = nlp(text)
for ent in doc.ents:
  print(f" entity : {ent.text}, label : {ent.label_}")
```

## Let's dive into Stemming and Lemmatization

Note : Stemmer does not have the knowledge of the language and just using fix rules. However, it s faster than Lem.

> spaCy does not support stemming so we use nltk

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["running", "ate", "eat", "went", "capable", "useful", "going", "better"]
for word in words:
  print(f" word : {word} |  stemmed word : {stemmer.stem(word)}")
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")  # differnt language models have differnt rules for lemmatization

text = " ".join(words)
doc = nlp(text)
for token in doc:
  print(f"word : {token} |  lemma : {token.lemma_} | hash : {token.lemma}")
```

````python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text  import CountVectorizer
````

````python
email_df = pd.read_csv("spam.csv")
email_df.head()
````

````python
email_df["Category"].value_counts()
# Our dataset is imbalance.
````

````python
# [1 if email == "ham" else 0 for email in email_df["Category"]]
email_df["Spam"] = email_df["Category"].apply(lambda email: 1 if email == "spam" else 0)
email_df.head()
````

````python
X = email_df["Message"]
y = email_df.Spam

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

print(f"dataset size is : {email_df.shape}")
print(f"train size is : {X_train.shape} and type of data is : {type(X_train)}")
print(f"test size is : {X_test.shape} ")

# X : Allowed inputs are lists, numpy arrays, scipy-sparse matrices or pandas dataframes.
````

To create Bag of words for numerical representation we need to call ```CountVectorizer```. It Convert a collection of text documents to a matrix of token counts. It will create vocabulary and for each of these emails it will have a word count at the given position. by using ```CountVectorizer```, we generate the following matrix and each of this vector is one unit.
![bag of words](/media/elham/Learning/NLP/bag_of_word.jpg)

````python
vectorizer = CountVectorizer()
X_train_cv = vectorizer.fit_transform(X_train)
X_train_cv
````

````python
type(X_train_cv)  #scipy.sparse._csr.csr_matrix
X_train_cv.toarray() # numpy array
````

# training

````python
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()
model.fit(X_train_cv, y_train)
````

**Note : The final output of fit(), transform(), and fit_transform() is going to be the same. Now, we will have to ensure that the same transformation is applied to the test dataset.  But, we cannot use the fit() method on the test dataset, because it will be the wrong approach as it could introduce bias to the testing dataset.**

# test

````python
X_test_cv = vectorizer.transform(X_test)
y_predict = model.predict(X_test_cv)
````

````python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_predict))
# if class is imbalance, we should consider F1-score.
````

````python
# test the result with new emails
emails = ["Up to 20% discount on parking, exclusive offer just for you. Do not miss this reward",
 "You recently requested an AWS service that required additional validation"]

emails_cv = vectorizer.transform(emails)

model.predict(emails_cv)
````

# pipline

````python
from sklearn.pipeline import Pipeline

clf = Pipeline([('vectorizer', CountVectorizer()), ('model', MultinomialNB())])

clf.fit(X_train, y_train)
````

````python
y_predict = clf.predict(X_test)
print(classification_report(y_test, y_predict))
````
